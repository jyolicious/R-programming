## 1. `print()` vs `cat()`

The `print()` is limited when trying to print multiple objects together, while `cat()` can combine multiple items into continuous output.

```r
# print() with one object
print("The zero occurs at")

# Multiple print() calls
print("The zero occurs at")
print(2*pi)
print("radians")

# cat() combines the values
cat("The zero occurs at", 2*pi, "radians.", "\n")
```

### `cat()` and newline

```r
d = date()
cat("Today's date is:", d, "\n")
```

### `cat()` with a separator

```r
x = 1:10

cat(x, sep = " ++ ")
cat("\n")

cat(x, sep = " / ")
cat("\n")
```

### `cat()` with calculations

```r
x = 7

cat("The square of", x, "is", x^2, "!\n")
cat("The square root of", x, "is approximately",
    format(sqrt(x), digits = 3), "\n")
```

### `cat()` with a vector

```r
evenno = c(2, 4, 6, 8, 10)

cat("The first few even numbers are:", evenno, "...\n")
```

### `cat()` with `fill` and `labels`

```r
x = 1:10

cat(x,
    fill = 2,
    labels = paste("(", letters[1:10], "):"))
```

## 2. `paste()` function

`paste()` creates a new character string by joining its arguments. Unlike `cat()`, its result can be assigned to a variable.

```r
paste("Everybody", "loves", "R Programming.")
paste("Everybody", "loves", "R Programming.", sep = "*")
paste("Everybody", "loves", "R Programming.", sep = "===")
```

### `paste()` with `sep`

```r
paste(1, " is first", 2, " is second", 3, " is third", sep = "#")

# Spaces inside the quoted strings are part of the result
paste(1, " is first ", 2, " is second ", 3, " is third ", sep = "#")
```

### `paste()` with vectors

```r
names = c("Prof. Singh", "Mr. Venkat", "Dr. Jha")

names

paste(names, "is", "a good", "person.")
```

### `collapse` parameter

```r
names = c("Prof. Singh", "Mr. Venkat", "Dr. Jha")

paste(names, "is", "a good", "person.", collapse = ", and ")
```

### `paste()` converts values to character strings

The notes show that `paste(1:12)` gives the character representation of the numbers, similar to `as.character(1:12)`.

```r
paste(1:12)
as.character(1:12)
```

### `paste()` and `paste0()` with vectors

```r
x = paste("Ex", 1:5, sep = "_")
x

x[1]
x[2]
x[3]
x[5]
```

```r
# With collapse, the result becomes one string
x = paste("Ex", 1:5, sep = "_", collapse = "")
x
x[1]
```

### Difference between `paste()` with and without `collapse`

```r
x1 = paste("Ex", 1:5, sep = "_")
x2 = paste("Ex", 1:5, sep = "_", collapse = "")

x1
x2
```

### `paste0()`

```r
paste0(1:10)
paste(1:10)

# paste0() does not insert the default space
paste0(1:10, c("st", "nd", "rd", rep("th", 7)))

# paste() inserts a space by default
paste(1:10, c("st", "nd", "rd", rep("th", 7)))
```

## 3. `strsplit()` — splitting strings

`strsplit()` splits the elements of a character vector according to the supplied separator.

```r
x = "The&!syntax&!of&!paste&!is!&available! &inthe online-help"

x

strsplit(x, split = "!")
```

### Splitting using `&!`

```r
x = "The&!syntax&!of&!paste&!is!&available! &inthe online-help"

strsplit(x, split = "&!")
```

### Accessing elements of the result

```r
x = "The&!syntax&!of&!paste&!is!&available! &inthe online-help"

y = strsplit(x, split = "!&")

y

# First component
y[[1]][1]

# Second component
y[[1]][2]

# Third component does not exist
y[[1]][3]
```

## 4. Convert split dates to a matrix

The notes use dates in the form `YYYY-MM-DD`, split them using `-`, and then convert the result into a matrix.

```r
dates = c("2020-07-24",
          "2021-08-25",
          "2022-09-26",
          "2023-10-27")

datesplt = strsplit(dates, "-")

datesplt
```

### Character matrix

```r
datemat = matrix(unlist(datesplt),
                 nrow = 4,
                 ncol = 3,
                 byrow = TRUE)

datemat
```

### Numeric matrix

```r
datematrix = matrix(as.numeric(unlist(datesplt)),
                   nrow = 4,
                   ncol = 3,
                   byrow = TRUE)

datematrix
```

### Splitting a word character-by-character

```r
strsplit("Shalabh", split = "")
```

## 5. `nchar()` — count characters

`nchar()` returns the number of characters in each element of a character vector.

```r
x = "R course 24.07.2022"
y = "Number of participants: 25"

nchar(x)
nchar(y)
```

### `nchar()` with a character vector

```r
x = c("Apple", "Banana", "Cake")
nchar(x)
```

### `nchar()` also works with numeric values by coercing them to character

```r
y = c(2, 4, 6)
nchar(y)

z = c(11, 222, 3333)
nchar(z)

z1 = c(1.1, 2.22, 3.333)
nchar(z1)
```

## 6. `nzchar()` — check for non-empty strings

`nzchar()` returns `TRUE` when a string is non-empty and `FALSE` for an empty string.

```r
x = "R course 24.07.2022"
y = "Number of participants: 25"

nzchar(x)
nzchar(y)
```

```r
x = c("Apple", "Banana", "Cake")
x
nzchar(x)

y = c("Apple", "", "Cake")
y
nzchar(y)
```

## 7. `tolower()` and `toupper()`

- `toupper()` converts alphabetic characters to uppercase.
- `tolower()` converts alphabetic characters to lowercase.
- Non-alphabetic characters remain unchanged.

```r
x = "R course will start from 24.07.2022"

x
toupper(x)

z = "INDIAN INSTITUTE OF TECHNOLOGY"

z
tolower(z)
```

## 9. Quick Revision

| Function | Main use |
|---|---|
| `print()` | Print an object |
| `cat()` | Print multiple items continuously |
| `paste()` | Join strings, with a default separator space |
| `paste0()` | Join strings without the default separator |
| `strsplit()` | Split strings |
| `nchar()` | Count characters |
| `nzchar()` | Check whether strings are non-empty |
| `toupper()` | Convert to uppercase |
| `tolower()` | Convert to lowercase |

### Important distinction

`sep` controls **what is placed between input arguments**.

`collapse` controls **how elements of the resulting vector are joined into one string**.
