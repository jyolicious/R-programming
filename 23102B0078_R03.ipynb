## Matrices

*Create a matrix*

```r
x <- matrix(
  nrow = 4,
  ncol = 2,
  data = c(1, 2, 3, 4, 5, 6, 7, 8)
)

x
```

*Access a Matrix Element*

```r
x[3, 2]
x[2, 1]
x[4, 2]
```

*Create Matrix Column-Wise Explicitly*

```r
x <- matrix(
  nrow = 4,
  ncol = 2,
  data = c(1, 2, 3, 4, 5, 6, 7, 8),
  byrow = FALSE
)

x
```

*Create Matrix Row-Wise*

```r
y <- matrix(
  nrow = 4,
  ncol = 2,
  data = c(1, 2, 3, 4, 5, 6, 7, 8),
  byrow = TRUE
)

y
```

*Matrix Calculations*

```r
x + x
x * 2
x^2
sum(x)
max(x)
min(x)
mean(x)
```

*Complete Matrix Example*

```r
# Create a 3 x 2 matrix
marks <- matrix(
  c(80, 75,
    90, 85,
    70, 95),
  nrow = 3,
  ncol = 2,
  byrow = TRUE
)

# Display matrix
marks

# Access an element
marks[2, 2]

# Matrix properties
dim(marks)
nrow(marks)
ncol(marks)
mode(marks)
attributes(marks)

# Mathematical operations
marks^2
marks * 2

# Statistical functions
sum(marks)
mean(marks)
max(marks)
min(marks)

# Check whether it is a matrix
is.matrix(marks)
```

*final demonstration*

```r
# Create vectors
a <- c(2, 3, 5, 7)
b <- c(1, 2, 3, 4)

# Arithmetic operations
addition <- a + b
subtraction <- a - b
multiplication <- a * b
division <- a / b

# Power
power <- a^2

# Integer division
integer_division <- a %/% 2

# Modulo
remainder <- a %% 2

# Built-in functions
maximum <- max(a)
minimum <- min(a)
average <- mean(a)
total <- sum(a)
product <- prod(a)

# Display results
addition
subtraction
multiplication
division
power
integer_division
remainder

maximum
minimum
average
total
product
```

*matrix final example*

```r
# Create a 4 x 2 matrix
M <- matrix(
  c(1, 2, 3, 4, 5, 6, 7, 8),
  nrow = 4,
  ncol = 2,
  byrow = TRUE
)

# Display matrix
M

# Access an element
M[3, 2]

# Matrix properties
dim(M)
nrow(M)
ncol(M)
mode(M)
attributes(M)

# Mathematical operations
M^2
M * 2

# Statistical functions
sum(M)
mean(M)
max(M)
min(M)
```
