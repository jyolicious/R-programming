# Part A – `for` Loop

The PDF defines the syntax as:

`for (name in vector) { commands to be executed }`

A `for` loop is useful when the number of repetitions is known in advance.

```r

%%R

# PDF Example 1
for (i in 1:5) {
  print(i^2)
}
```

```r
%%R

# PDF Example 2
for (i in c(2, 4, 6, 7)) {
  print(i^2)
}
```

# Part B – `for` Loop with a Function

This reproduces the PDF's example that counts values satisfying `xval/2 > 3`.

```r
%%R

x <- c(2, 4, 6, 8, 10, 12)

excount <- function(x) {
  count <- 0

  for (xval in x) {
    if (xval / 2 > 3) {
      count <- count + 1
    }
  }

  print(count)
}

excount(x)
```

# Part C – Nested `for` Loops

The PDF demonstrates a loop inside another loop and uses `paste()` to combine values.

```r
%%R

child <- c("child1", "child2", "child3")
sweet <- c("sweet1", "sweet2", "sweet3")

for (x in child) {
  for (y in sweet) {
    print(paste(x, y))
  }
}
```

# Part D – `break`

The PDF explains that `break` stops the loop before all items have been processed.

```r
%%R

drink <- c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "tea") {
    break
  }
  print(x)
}
```

### Practice 3
Stop a loop when the number 7 is reached.

```r
%%R

for (i in 1:10) {
  if (i == 7) {
    break
  }
  print(i)
}
```

# Part E – `next`

The PDF explains that `next` skips the current iteration without terminating the loop.

```r
%%R

drink <- c("coffee", "lemonade", "tea", "juice")

for (x in drink) {
  if (x == "lemonade") {
    next
  }
  print(x)
}
```

### Practice 4

```r
%%R

for (i in 1:10) {
  if (i %% 2 == 0) {
    next
  }
  print(i)
}
```

# Part F – `while` Loop

According to the PDF, a `while` loop is useful when the number of repetitions is not known in advance.

Syntax:

`while(condition) { commands }`

```r
%%R

# PDF Example 1
i <- 1

while (i < 10) {
  print(i^2)
  i <- i + 2
}
```

### Practice 5

## PDF Example – Sum using `while`

The original PDF uses `readline()` to ask the user for a number less than 25.  
For Colab, the value is supplied directly so the cell can run without interactive console input.

```r
%%R

sumfunction <- function(number) {
  sum_value <- 0

  while (number <= 25) {
    sum_value <- sum_value + number
    number <- number + 1
  }

  print(paste(
    "The sum of numbers received from the While Loop:",
    sum_value
  ))
}

sumfunction(22)
```

# Part G – `repeat` Loop

The PDF states that `repeat` does not test a condition before entering the loop.  
The programmer must explicitly terminate it, usually with `break`.

Syntax:

`repeat { commands }`

```r
%%R

# PDF Example 3
i <- 1

repeat {
  print(i^2)
  i <- i + 2

  if (i > 10) {
    break
  }
}
```

### `repeat` with `next` and `break`

This follows the structure of the PDF's Example 4.

```r
%%R

i <- 1

repeat {
  i <- i + 1

  if (i < 10) {
    next
  }

  print(i^2)

  if (i >= 13) {
    break
  }
}
```

# Part H – Functions

The PDF describes functions as groups of commands that take input arguments, perform calculations, and return a result.

General syntax:

`name <- function(argument1, argument2, ...) { expression(s) }`

## Function with one argument

```r
%%R

abc <- function(x) {
  x^2
}

abc(3)
abc(6)
abc(9)
```

## Function with two arguments

```r
%%R

abc <- function(x, y) {
  x^2 + y^2
}

abc(3, 4)
abc(10, 10)
abc(-2, -3)
```

## Function using `sin()` and `cos()`

This reproduces the calculation shown in the PDF.

```r
%%R

abc <- function(x) {
  sin(x)^2 + cos(x)^2 + x
}

abc(9)
abc(99)
abc(-15)
```

## Function without arguments

The PDF also demonstrates a function that contains a `for` loop but takes no arguments.

```r
%%R

abc <- function() {
  for (i in 1:3) {
    print(i^3)
  }
}

abc()
```

### Practice 6
Create a function that receives a number and returns its factorial using a `for` loop.

```r
%%R

factorial_for <- function(n) {
  result <- 1

  for (i in 1:n) {
    result <- result * i
  }

  return(result)
}

factorial_for(5)
factorial_for(7)
```

# Part I – Sequences with `seq()`

The PDF introduces `seq()` for generating regular sequences.

General form:

`seq(from, to, by, length.out, along.with, ...)`

```r
%%R

# Default increment
seq(from = 2, to = 4)
seq(from = 4, to = 2)
seq(from = -4, to = 4)
```

## Constant increment

```r
%%R

# PDF example
seq(from = 10, to = 20, by = 2)
```

## Constant decrement

```r
%%R

# PDF example
seq(from = 20, to = 10, by = -2)
```

## Fractional decrement

```r
%%R

# PDF example
seq(from = 3, to = -2, by = -0.5)
```

## Sequence with predefined length

```r
%%R

seq(to = 10, length.out = 10)
seq(from = 10, length.out = 10)
```

## Fractional increment with predefined length

```r
%%R

seq(from = 10, length.out = 10, by = 0.1)
```

## Decrement with predefined length

```r
%%R

seq(from = 10, length.out = 10, by = -2)
```

## Fractional decrement with predefined length

```r
%%R

seq(from = 10, length.out = 5, by = -0.2)
```

### Exercise 4
Write a function that accepts a vector and counts how many values are greater than 5.

```r
%%R

count_greater_than_5 <- function(x) {
  count <- 0

  for (value in x) {
    if (value > 5) {
      count <- count + 1
    }
  }

  return(count)
}

x <- c(2, 4, 6, 8, 10, 3, 1, 7)
count_greater_than_5(x)
```

# Quick Revision

| Topic | Main idea | Key syntax |
|---|---|---|
| `for` | Known/repeated values | `for (i in vector)` |
| Nested `for` | Loop inside a loop | `for (...) { for (...) { } }` |
| `break` | Stop the loop | `break` |
| `next` | Skip current iteration | `next` |
| `while` | Repeat while condition is TRUE | `while(condition)` |
| `repeat` | Repeat until explicitly stopped | `repeat { ... break }` |
| Function | Reusable block of commands | `f <- function(x) { ... }` |
| `seq()` | Generate regular sequences | `seq(from, to, by)` |
