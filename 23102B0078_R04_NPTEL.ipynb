## Lecture 13 – Matrix Operations: Rows, Columns and Other Operations

Topics: creating matrices, renaming rows/columns with `rownames()` and `colnames()`, assigning a constant to all elements, diagonal matrices with `diag()`, transpose with `t()`, row/column sums with `rowSums()` and `colSums()`, and row/column means with `rowMeans()` and `colMeans()`.

```r
# Matrix creation and row/column names
x <- matrix(nrow = 4, ncol = 3, data = 1:12)
print(x)

rownames(x) <- c("r1", "r2", "r3", "r4")
colnames(x) <- c("c1", "c2", "c3")
print(x)

# Assign one value to every matrix element
constant_matrix <- matrix(nrow = 4, ncol = 2, data = 2)
print(constant_matrix)

# Diagonal matrices
d1 <- diag(1, nrow = 3, ncol = 3)
d5 <- diag(5, nrow = 3, ncol = 3)
print(d1)
print(d5)

# Transpose
m <- matrix(nrow = 4, ncol = 2, data = 1:8, byrow = TRUE)
print(m)
print(t(m))

# Row and column sums / means
print(rowSums(m))
print(colSums(m))
print(rowMeans(m))
print(colMeans(m))
```

## Lecture 14 – Matrix Access and Mathematical Operations

Matrix elements are accessed using `[row, column]`. A complete row can be selected with `x[3,]`, a column with `x[,2]`, and a submatrix with ranges such as `x[4:5, 2:3]` or selected indices such as `x[c(1,4), c(1,3)]`.

The notes also show scalar addition, subtraction, multiplication and division, where the operation is applied to every element.

```r
# Create matrix row-wise
x <- matrix(nrow = 5, ncol = 3, byrow = TRUE, data = 1:15)
print(x)

# Access rows, columns and submatrices
print(x[3, ])
print(x[, 2])
print(x[4:5, 2:3])
print(x[c(1, 4), c(1, 3)])

# Scalar operations
m <- matrix(nrow = 4, ncol = 2, data = 1:8, byrow = TRUE)
print(m + 5)
print(m - 5)
print(5 * m)
print(m / 2)

# Matrix addition and subtraction (same dimensions)
y <- matrix(nrow = 4, ncol = 2, data = 11:18, byrow = TRUE)
print(m + y)
print(m - y)
```

## Lecture 15 – Matrix Mathematical Operations

Important operations from the notes:

- Matrix multiplication: `%*%`
- Transpose: `t()`
- Cross-product: `crossprod(x)` for `t(x) %*% x`
- Row-wise concatenation: `rbind()`
- Column-wise concatenation: `cbind()`
- Matrix inverse: `solve()`
- Eigenvalues and eigenvectors: `eigen()`

```r
# Matrix multiplication
x <- matrix(nrow = 4, ncol = 2, data = 1:8, byrow = TRUE)
y <- matrix(nrow = 2, ncol = 4, data = 11:18, byrow = TRUE)
print(x %*% y)
print(y %*% x)

# X'X and XX'
print(t(x) %*% x)
print(x %*% t(x))
print(crossprod(x))

# Concatenating matrices
a <- matrix(nrow = 3, ncol = 2, data = 1:6, byrow = TRUE)
b <- matrix(nrow = 3, ncol = 2, data = 11:16, byrow = TRUE)
print(rbind(a, b))
print(cbind(a, b))

# Inverse of a matrix
z <- matrix(nrow = 2, ncol = 2, byrow = TRUE,
           data = c(84, 100, 100, 120))
print(solve(z))

# Eigenvalues and eigenvectors
print(eigen(z))
```

## Lectures 16–17 – Logical and Relational Operators

The notes cover comparisons using `>`, `>=`, `<`, `<=`, `==`, `!=`, and `!`, together with logical operators `&`, `&&`, `|`, `||`, `xor()`, `isTRUE()` and `isFALSE()`.

**Important distinction:** `&` and `|` work element-wise on vectors, while `&&` and `||` examine only the first element and evaluate left-to-right. The notes also demonstrate filtering a vector using a logical condition.

```r
# Basic relational operators
print(8 > 7)
print(7 < 5)
print(7 >= 7)
print(8 <= 8)
print(8 != 9)
print(7 == 7)

# Negation
x <- TRUE
print(!x)

# Vector comparisons are element-wise
x <- c(1, 2, 3)
y <- c(4, 5, 6)
print(x > y)
print(x < y)
print(x != y)
print(x == y)

# AND / OR with scalar values
x <- TRUE
y <- FALSE
print(x & y)
print(x | y)
print(!x)
print(xor(x, y))

# & and | operate element-wise on vectors
v <- 1:6
print((v > 2) & (v < 5))
print(v[(v > 2) & (v < 5)])
print((v > 2) | (v < 5))
print(v[(v > 2) | (v < 5)])

# && and || examine only the first element
v <- c(8, 18)
print((v < 10) || (v < 2))
print((v < 10) | (v < 2))
print((v < 10) && (v > 2))
print((v < 10) & (v > 2))

# isTRUE(), isFALSE(), and is.logical()
Logical1 <- (5 > 2)
Logical2 <- (5 < 10)
Logical3 <- (5 != 5)
print(Logical1)
print(is.logical(Logical1))
print(Logical2)
print(is.logical(Logical2))
print(Logical3)
print(is.logical(Logical3))

print(isTRUE(8 < 6))
print(isTRUE(8 > 6))
print(isFALSE(5 < 8))
print(isFALSE(5 > 8))
```
