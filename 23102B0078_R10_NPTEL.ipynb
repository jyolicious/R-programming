## 1. `sub()` and `gsub()`

`sub()` replaces the **first** matching occurrence. `gsub()` replaces **all** matching occurrences.

```r
y = "Number of participants: 25"

sub("25", "30", y)
```

```r
y = "Mr. Singh is the smart one. Mr. Singh is funny, too."

# Replace only the first occurrence
sub("Mr. Singh", "Professor Jha", y)

# Replace every occurrence
gsub("Mr. Singh", "Professor Jha", y)
```

## 2. `grep()`

`grep()` searches a character vector and returns the indices of elements containing the pattern. With `value = TRUE`, it returns the matching strings.

```r
str = c(
  "R Course",
  "exercises",
  "include examples of R language"
)

grep("ex", str, value = TRUE)
grep("ex", str, value = FALSE)
```

### `ignore.case` in `grep()`

```r
str = c(
  "R Course",
  "exercises",
  "include examples of r language",
  "in R software."
)

grep("R", str, ignore.case = FALSE, value = TRUE)
grep("R", str, ignore.case = TRUE, value = TRUE)
grep("R", str, ignore.case = TRUE, value = FALSE)
grep("R", str, ignore.case = FALSE, value = FALSE)
```

### More `grep()` examples

```r
x = "R course 24.07.2022"
y = "Number of participants: 25"

c(x, y)

grep("our", c(x, y))
grep("Num", c(x, y))
```

## 3. `grepl()`

`grepl()` returns `TRUE` or `FALSE` for each element, indicating whether the pattern occurs.

```r
str = c(
  "R Course",
  "exercises",
  "include examples of R language"
)

grepl("R", str)
grepl("ex", str)
```

## 4. Data Frames

A data frame stores variables of equal length as columns and observations as rows. Unlike a matrix, a data frame can contain different types of variables.

```r
students = data.frame(
  Name = c("Asha", "Riya", "Neha", "Pooja"),
  Marks = c(85, 92, 78, 88),
  Branch = c("CMPN", "IT", "CMPN", "AI")
)

students
```

### Accessing columns

```r
students$Name
students$Marks
students$Branch
```

## 10. Combining Data Frames

The notes introduce three main operations:

- `cbind()` — combine columns side-by-side
- `merge()` — join using a common column
- `rbind()` — stack rows vertically

```r
df1 = data.frame(
  state = c("UP", "MP", "AP", "JK"),
  popnsize = c(1000, 2000, 3000, 4000)
)

df2 = data.frame(
  state = c("UP", "MP", "AP", "JK"),
  samplesize = c(100, 200, 300, 400),
  surveycompleted = c("Yes", "No", "Yes", "No")
)

df1
df2
```

## 11. `cbind()` — horizontal combination

```r
cbind(df1, df2)
```

## 12. `merge()` — combine using a common column

Both data frames have `state` in common, so we merge using `by = "state"`.

```r
merge(df1, df2, by = "state")
```

## 13. `rbind()` — vertical combination

```r
df11 = data.frame(
  state = c("UP", "MP", "AP", "JK"),
  popnsize = c(1000, 2000, 3000, 4000)
)

df22 = data.frame(
  state = c("Bihar", "Delhi", "Punjab"),
  popnsize = c(100, 200, 300)
)

df11
df22
```

```r
rbind(df11, df22)
```

## 14. Practice Questions

1. Replace the first `"R"` in a string using `sub()`.
2. Replace every `"R"` using `gsub()`.
3. Use `grep()` to find strings containing `"data"`.
4. Use `grepl()` to return TRUE/FALSE for the same search.
5. Create a data frame containing Student, Marks and Branch.
6. Select students with Marks greater than 80 using `subset()`.
7. Split a data frame according to Branch.
8. Create two data frames and combine them with `cbind()`.
9. Merge two data frames using a common ID column.
10. Stack two compatible data frames using `rbind()`.


```r
# Practice 1
s = "R is useful. R is popular."
sub("R", "Python", s)

# Practice 2
gsub("R", "Python", s)

# Practice 3
words = c("data science", "machine learning", "data frame", "statistics")
grep("data", words, value = TRUE)

# Practice 4
grepl("data", words)

# Practice 5
student_df = data.frame(
  Student = c("A", "B", "C", "D"),
  Marks = c(75, 91, 84, 68),
  Branch = c("CMPN", "IT", "CMPN", "AI")
)
student_df

# Practice 6
subset(student_df, Marks > 80)

# Practice 7
split(student_df, student_df$Branch)

# Practice 8
extra = data.frame(Attendance = c(90, 95, 88, 80))
cbind(student_df, extra)

# Practice 9
id1 = data.frame(ID = c(1,2,3), Name = c("A","B","C"))
id2 = data.frame(ID = c(1,2,3), Score = c(80,90,85))
merge(id1, id2, by = "ID")

# Practice 10
rbind(
  data.frame(ID = c(4,5), Name = c("D","E")),
  data.frame(ID = c(6,7), Name = c("F","G"))
)
```
