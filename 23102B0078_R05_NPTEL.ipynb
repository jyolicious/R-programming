## Lecture 18 – Missing Data

R represents missing observations using `NA`. Important functions from the notes:

- `is.na()` – detects missing values
- `which(is.na(x))` – gives positions of missing values
- `sum(is.na(x))` – counts missing values
- `complete.cases(x)` – identifies complete observations
- `na.omit(x)` – removes missing observations
- `mean(x, na.rm = TRUE)` – calculates the mean while removing `NA`

**Note:** `NA` and `NULL` are different. `NA` represents something that exists but is missing, whereas `NULL` represents something that never existed.

```r
# Missing data examples
x <- NA
print(x)
print(is.na(x))

x <- c(11, NA, 13, NA)
print(x)
print(is.na(x))

# Mean with and without removing NA
print(mean(x))
print(mean(x, na.rm = TRUE))

# Positions and count of missing values
print(which(is.na(x)))
print(sum(is.na(x)))

# Complete cases and removal of missing values
print(complete.cases(x))
y <- na.omit(x)
print(y)
print(mean(y))
```

## Lecture 19 – Conditional Execution: `if` and `if-else`

`if` executes a block only when the condition is `TRUE`. `if-else` provides two alternatives. The notes emphasize using these for a **single-element condition**, rather than a vector-valued condition.

The logical operators `&&` (AND) and `||` (OR) can be used in complex conditions.

```r
# if examples
x <- 5
if (x > 4) {
  print(x * 3)
}

x <- 3
if (x > 4) {
  print(x * 3)
}

x <- 6
if (x > 3) {
  print("The value is more than 3")
}

# if-else examples
x <- 5
if (x == 3) {
  x <- x - 1
} else {
  x <- 2 * x
}
print(x)

x <- 2
if (x > 3) {
  print("The value is more than 3")
} else {
  print("The value is less than 3")
}
```

## Lecture 20 – Nested `if-else` and `ifelse()`

`if ... else if ... else` is used when there are more than two alternatives.

`ifelse(test, yes, no)` is different: it supports **vector-valued conditions**, applying `yes` where the test is `TRUE` and `no` where it is `FALSE`.

```r
# Nested if-else examples
x <- 5
if (x == 3) {
  x <- x - 1
} else if (x < 3) {
  x <- x + 5
} else {
  x <- 2 * x
}
print(x)

x <- 2
if (x == 3) {
  x <- x - 1
} else if (x < 3) {
  x <- x + 5
} else {
  x <- 2 * x
}
print(x)

x <- 3
if (x == 3) {
  x <- x - 1
} else if (x < 3) {
  x <- x + 5
} else {
  x <- 2 * x
}
print(x)

# ifelse() with a vector
x <- 1:10
print(ifelse(x < 6, x^2, x + 1))

# Even / odd example from the notes
x <- c(7, 9, 8, 4)
print(ifelse(x %% 2 == 0, "even number", "odd number"))
```

## Lecture 21 – `switch()` and `which()`

`switch()` is a multiway branch statement and can select a value using either an integer or a character expression.

`which()` returns the positions where a logical condition is `TRUE`. For matrices, `arr.ind = TRUE` returns row and column positions. The notes also demonstrate `which.min()` and `which.max()`.

```r
# switch() using an integer
print(switch(2, "apple", "banana", "orange"))
print(switch(1, "apple", "banana", "orange"))

# switch() using a character string
print(switch("colour", colour = "blue", gender = "male", volume = 50))
print(switch("volume", colour = "blue", gender = "male", volume = 50))

# No-match examples return NULL
print(switch(4, "apple", "banana", "orange"))
print(switch("size", colour = "blue", gender = "male", volume = 50))

# which() examples
x <- c(10, 15, 8, 14, 6, 12)
print(x)
print(which(x == 14))
print(which(x != 12))
print(which(x > 10))

# which.min() and which.max() on a matrix
m <- matrix(nrow = 3, ncol = 3, data = 1:9)
print(m)
print(which.min(m))
print(which.max(m))

# which() on matrix values
print(which(m %% 2 == 1))
print(which(m %% 2 == 1, arr.ind = TRUE))
```
