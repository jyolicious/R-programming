## 1. Sequences

The notes introduce `seq()` for generating regular sequences. The `:` operator can also generate sequences with a unit increment/decrement.

```r
# Continuous sequences with unit increment
1:10
10:1
5:15
15:5

# Negative sequences
-1:-10
-10:-1
-5:-15
-15:-5

# Decimal sequences using :
1.23:10
1.23:10.54
10.54:2.23
-1.23:-10
-5.23:6
```

### Using `seq()`

```r
# Basic seq()
seq(10)
seq(1:10)

# Sequence with a predefined variable and constant increment
x <- 2
seq(1, x, x/10)

x <- 50
seq(0, x, x/10)

# Store a sequence and perform an operation
x <- seq(1, 50, 1/2)
x

y <- 2 * x
y
```

### Index vector

```r
x <- c(9, 8, 7, 6)
ind <- seq(along = x)

ind
x[ind[2]]
```

## 2. Date Sequences

The notes use `as.Date()` together with `seq()` to generate sequences by days, months and years. `Sys.time()` gives the current date and time, while `Sys.Date()` gives the current date.

```r
# Current date and time
Sys.time()

# Current date
Sys.Date()

# Sequence of first day of years
seq(
  as.Date("2010-01-01"),
  as.Date("2017-01-01"),
  by = "years"
)

# Sequence of days
seq(
  as.Date("2017-01-01"),
  by = "days",
  length.out = 6
)

# Sequence of months
seq(
  as.Date("2017-01-01"),
  by = "months",
  length.out = 6
)

# Sequence of years
seq(
  as.Date("2017-01-01"),
  by = "years",
  length.out = 6
)
```

### Date sequence with a defined start and end date

```r
startdate <- as.Date("2016-01-01")
enddate <- as.Date("2017-01-01")

out <- seq(enddate, startdate, by = "-1 month")
out
```

## 4. Repeats using `rep()`

The `rep()` function replicates numeric values, character values, or vector elements.

Main forms:
- `rep(x, times = n)` → repeats the complete object
- `rep(x, each = n)` → repeats each element
- `rep(x, length.out = n)` / `rep(x, length = n)` → produces a specified length

```r
# Repeat a value
rep(3.5, times = 10)

# Repeat a complete vector
rep(1:4, times = 2)

# Repeat every element
x <- 1:4
rep(x, times = 3)
rep(x, each = 3)

# each and times together
rep(1:4, each = 2)
rep(1:4, each = 2, times = 3)
rep(1:4, times = 3, each = 2)
```

### Different number of repetitions for each element

```r
# Elements 1, 2, 3 and 4 are repeated 2, 3, 4 and 5 times
rep(1:4, 2:5)

# Generate the repetition counts using seq()
ans <- seq(from = 2, to = 8, by = 2)
ans

rep(1:4, ans)
```

### Repeating matrices and character values

```r
# Matrix repetition
x <- matrix(nrow = 2, ncol = 2, data = 1:4, byrow = TRUE)
x
rep(x, 3)

# Character repetition
rep(c("a", "b", "c"), 2)
rep(c("apple", "banana", "cake"), 2)

# Pre-specified output length
rep(2, length.out = 5)
rep(2, length = 5)
rep(c(2, 3), length = 5)
rep(c(2, 3, 4), length = 5)

rep("apple", length = 5)
rep(c("a", "b", "c"), length = 2)
rep(c("a", "b", "c"), length = 5)
```

## 5. Sorting

`sort()` sorts a vector in ascending order by default. Set `decreasing = TRUE` for descending order.

```r
y <- c(8, 5, 7, 6)

y
sort(y)
sort(y, decreasing = TRUE)

# Sorting with missing values
z <- c(8, NA, 5, 7, 6)

sort(z, na.last = TRUE)   # NA last
sort(z, na.last = FALSE)  # NA first
sort(z, na.last = NA)     # remove NA
```

## 6. Ordering

`order()` returns the **positions/indices** that would arrange a vector in sorted order.

For `y = c(9, 8, 5, 7, 6)`, the ascending order of values is `5, 6, 7, 8, 9`, whose original positions are `3, 5, 4, 2, 1`.

```r
y <- c(9, 8, 5, 7, 6)

y
order(y)
order(y, decreasing = TRUE)

# Use order() to actually rearrange the vector
y[order(y)]
y[order(y, decreasing = TRUE)]
```

## 7. Mode

The `mode()` function indicates how an R object is stored. The notes demonstrate numeric, character, list and function modes.

```r
mode(2.432)
mode(c(3, 4, 5, 6, 7, 8))
mode("India")
mode(c("India", "CANADA"))

mode(factor(c("UP", "MP")))
mode(list("India", "USA"))

df <- data.frame(
  x = 1:2,
  y = c("India", "USA")
)
mode(df)

mode(print)
```

## 8. Lists

Unlike vectors, lists can contain objects of different types. The notes demonstrate lists containing character vectors, numeric vectors and matrices. Lists can also contain other structured objects.

```r
# A list with different object types
z1 <- list(
  c("water", "juice", "lemonade"),
  rep(1:4, each = 2),
  matrix(data = 5:8, nrow = 2, ncol = 2, byrow = TRUE)
)

z1
```

### Accessing list elements with `[[ ]]`

```r
# First element of the list
z1[[1]]

# Extract "juice"
z1[[1]][2]

# Compare [ ] and [[ ]]
z1[1]
z1[[1]]

# The notes specifically demonstrate that:
# z1[1][2] does NOT extract "juice"
z1[1][2]

# Correct extraction:
z1[[1]][2]
```

### Lists containing matrices

```r
x1 <- matrix(nrow = 2, ncol = 2, data = 1:4, byrow = TRUE)
x2 <- matrix(nrow = 2, ncol = 2, data = 5:8, byrow = TRUE)

x1
x2
x1 + x2

# Store matrices in a list
matlist <- list(x1, x2)
matlist

matlist[1]
matlist[2]

# Extract the matrix itself
matlist[[1]]
matlist[[2]]
```

### Named list elements

```r
student_data <- list(
  Students = c("Asha", "Ravi", "Meena"),
  Marks = c(85, 92, 78),
  Passed = c(TRUE, TRUE, TRUE)
)

student_data

# Access by name
student_data[["Students"]]
student_data$Students

# Access by position
student_data[[2]]

# Extract a sublist
student_data[c(1, 3)]
```
