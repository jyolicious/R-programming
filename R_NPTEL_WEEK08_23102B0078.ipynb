## 1. Operations on Lists

The notes demonstrate merging lists with `c()`, converting lists to vectors with `unlist()`, appending with `append()`, removing elements using negative indexes, and extracting elements using indexes.

```r
# Create two lists
list1 <- list(1, 2, 3)
list2 <- list("water", "juice", "lemonade")

list1
list2
```

```r
# Merge two lists
list12 <- c(list1, list2)
list12
```

```r
# Convert a list to a vector
unlist(list1)
unlist(list2)

# Check modes
mode(list1)
mode(unlist(list1))
```

```r
# Append an element to a list
append(list1, 100)
append(list2, "coffee")
```

```r
# Append after a particular position
append(list1, 100, after = 2)
append(list2, "coffee", after = 2)
```

```r
# Remove elements using negative indexing
list1[-2]
list2[-1]
```

```r
# Extract a range of elements
list3 <- list(1, 2, 3, 4, 5, 6)

list3[2:4]
list3[c(1, 3, 5)]
```

```r
# Extract elements from a character list
drinks <- list("water", "juice", "lemonade", "tea", "coffee", "milk")

drinks[2:4]
drinks[c(1, 3, 5)]
```

## 2. Vector Indexing

The notes demonstrate:
- logical indexing,
- filtering values using conditions,
- the modulo operator `%%`,
- handling missing observations with `NA`,
- negative integer indexing,
- named vectors,
- and an empty index.

```r
# Basic vector indexing
x <- 1:10
x

# Values greater than 5
x[x > 5]

# Even values: %% gives the remainder
x[x %% 2 == 0]

# Odd values
x[x %% 2 == 1]
```

```r
# Missing observation
x <- 1:10
x[5] <- NA

x

# Remove missing values using !is.na()
y <- x[!is.na(x)]
y

# Mean with NA versus mean after removing NA
mean(x)
mean(y)
```

```r
# Negative indexing
x <- 1:10

x[-(1:5)]
x[6:10]
```

```r
# Named vector
scores <- c(water = 1, juice = 2, lemonade = 3)

names(scores)
scores["juice"]
```

```r
# Empty index returns the complete vector
x <- 1:10
x[]
```

```r
# Named list and changing a name
z <- list(a1 = 1, a2 = "c", a3 = 1:3)

names(z)

names(z)[3] <- "c2"
z
```

## 3. Factors

According to the notes, factors represent **categorical variables** and are used as **grouping indicators**. Each possible value is called a **level**. Factors are internally represented using numeric codes with labels.

The `factor()` function can specify `levels`, `labels`, and `exclude`.

```r
# Basic factor
gender <- factor(c("Male", "Female", "Female", "Male", "Female"))

gender
levels(gender)
```

```r
# Explicit levels and labels using the die example
y <- c(1, 4, 3, 5, 4, 2, 4)

possible.dieface <- c(1, 2, 3, 4, 5, 6)
labels.dieface <- c("one", "two", "three", "four", "five", "six")

facy <- factor(
    y,
    levels = possible.dieface,
    labels = labels.dieface
)

facy
levels(facy)
```

```r
# as.factor()
x <- c(3, 4, 5, 6, 1, 2, 3, 3, 4, 4, 5, 6)

y <- as.factor(x)

y
levels(y)
```

```r
# Character values become factor levels
x <- factor(c("lemonade", "lemonade", "juice", "lemonade", "water"))

x
levels(x)
```

## 4. `class()` and `unclass()`

The notes explain that `class()` reports the class of an R object. `unclass()` temporarily removes the effects of the class.

For factors, `unclass()` exposes the integer codes assigned to the factor levels.

```r
# class()
class(9)
class("9")
class(print)

m <- matrix(nrow = 2, ncol = 2, data = 1:4)
class(m)
```

```r
# Factor and its internal codes
brands <- c("A", "A", "B", "B", "B", "B", "C")
brands_fac <- factor(brands)

brands_fac
unclass(brands_fac)
```

```r
# Use the factor codes to index another vector
colours <- c("blue", "green", "red")

colours[unclass(brands_fac)]
```

```r
# Explicit factor level order
x <- factor(
    c("lemonade", "lemonade", "juice", "lemonade", "water"),
    levels = c("water", "juice", "lemonade")
)

x
unclass(x)
levels(x)
```

```r
# Ordered factor
income <- ordered(
    c("high", "high", "low", "medium", "medium"),
    levels = c("low", "medium", "high")
)

income
unclass(income)
levels(income)
```

## 5. Strings – Display and Formatting

The notes introduce the formatting/display commands:
- `print()`
- `format()`
- `cat()`
- `paste()`

The examples below focus on the documented `print()` and `format()` functionality, including `digits`, `nsmall`, `width`, `justify`, and `big.mark`.

```r
# print() with numbers
print(sqrt(2))
print(sqrt(2), digits = 5)
print(sqrt(2), digits = 10)
```

```r
# print() with character vectors
print("apple")
print(c("apple", "banana"))
print(c("apple", "banana", 6, 10))
```

```r
# format(): digits and nsmall
print(format(0.5, digits = 10, nsmall = 15))
```

```r
# format(): width and centre justification
format(
    c("A", "BB", "CCC", "DDDD"),
    width = 7,
    justify = "centre"
)

format(
    c("A", "BB", "CCC", "DDDD"),
    width = 14,
    justify = "centre"
)
```

```r
# format(): different justification options
text <- c("A", "BB", "CCC", "DDDD")

format(text, width = 7, justify = "centre")
format(text, width = 7, justify = "left")
format(text, width = 7, justify = "right")
format(text, width = 7, justify = "none")
```

```r
# Formatting a matrix
x <- matrix(
    nrow = 3,
    ncol = 2,
    data = 1:6,
    byrow = TRUE
)

print(x)
```

```r
# Formatting large quantities with big.mark
format(1234567, big.mark = ",")
format(12345678, big.mark = ",")
format(123456789, big.mark = ",")

format(123456789, big.mark = " ")
```

## 6. Integrated Mini Exercise

Use lists, vector indexing, factors, and formatting together.

**Task:** Create a small set of student records containing:
- student names,
- marks,
- and performance categories.

Then:
1. Identify students scoring above 75.
2. Convert performance into a factor.
3. Inspect its levels.
4. Display a formatted summary.

```r
# Integrated example
students <- c("Asha", "Riya", "Neha", "Karan", "Vivek")
marks <- c(72, 88, 91, 64, 79)

performance <- c("Good", "Excellent", "Excellent", "Average", "Good")
performance_fac <- factor(performance)

# Students scoring above 75
students[marks > 75]

# Corresponding marks
marks[marks > 75]

# Factor representation
performance_fac
levels(performance_fac)

# Formatted marks
format(marks, nsmall = 2)
```
